```python
q = ["cold", "hot"]
N = len(q)

# a[i][j] probability of moving from state i to state j
a = [
    [0.5, 0.5],
    [0.4, 0.6]
]

# b[i][j] probability of getting observation j knowing state i happened
b = [
    {"1": 0.5, "2": 0.4, "3": 0.1},
    {"1": 0.2, "2": 0.4, "3": 0.4}
]

pi = [0.2, 0.8]
```

```python
def print_2d_matrix(lst):
    for row in lst:
       print([round(num, 8) for num in row])
```

```python
o = ["2", "1", "1", "3"]
T = len(o)

alpha = [[0 for _ in range(T)] for _ in range(N)]

print_2d_matrix(alpha)
```

```python
for index in range(len(pi)):
    alpha[index][0] = pi[index] * b[index][o[0]]

print_2d_matrix(alpha)
```

```python
for t in range(1, T):
    observation = o[t]
    for state in range(0, N):
        for state_prev in range(0, N):
            alpha[state][t] += alpha[state_prev][t - 1] * a[state_prev][state] * b[state][observation]
```

```python
print_2d_matrix(alpha)
```

```python
probability = sum(alpha[state][-1] for state in range(len(q)))
print(round(probability, 8))
```
