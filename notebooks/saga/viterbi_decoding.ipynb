```python
q = ["cold", "hot"]
N = len(q)

# a[i][j] probability of moving from state i to state j
a = [
    [0.5, 0.5],
    [0.4, 0.6]
]

# b[i][j] probability of getting observation j knowing state i happened
b = [
    {"1": 0.5, "2": 0.4, "3": 0.1},
    {"1": 0.2, "2": 0.4, "3": 0.4}
]

pi = [0.2, 0.8]
```

```python
def print_2d_matrix(lst):
    for row in lst:
       print([round(num, 8) for num in row])
```

```python
o = ["2", "1", "1", "3"]
T = len(o)

viterbi = [[0 for _ in range(T)] for _ in range(N)]
backpointer = [[0 for _ in range(T)] for _ in range(N)]

print_2d_matrix(viterbi)
```

```python
for index in range(N):
    viterbi[index][0] = pi[index] * b[index][o[0]]

print_2d_matrix(viterbi)
```

```python
for t in range(1, T):
    observation = o[t]
    for state in range(0, N):
        for state_prev in range(0, N):
            max_value = viterbi[state_prev][t - 1] * a[state_prev][state] * b[state][observation]
            if (max_value > viterbi[state][t]):
                viterbi[state][t] = max_value
                backpointer[state][t] = state_prev
```

```python
print_2d_matrix(viterbi)
print_2d_matrix(backpointer)
```

```python
bestpathprob = -1
bestpathpointer = -1

for i in range(0, N):
    bestpathprob = max(bestpathprob, viterbi[i][-1])

print(round(bestpathprob, 8))
```

```python
for i in range(0, N):
    if viterbi[i][-1] == bestpathprob:
        bestpathpointer = i

print(bestpathpointer)
```

```python
bestpath = [bestpathpointer]

for t in range(T - 1, 0, -1):
    bestpath.append(backpointer[bestpath[-1]][t])

bestpath = list(reversed(bestpath))
print(bestpath)
```

```python
for i in range(T):
    print(q[bestpath[i]], end=" ")
```
